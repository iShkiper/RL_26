# Сорвенование по LunarLander

Необходимо обучить агента управлять лунным модулем в среде LunarLander-v3 в усложнённых условиях: действует постоянный боковой ветер и случайные вращательные возмущения — турбулентность. Гравитация также немного усилена. Эпизод ограничен 600 шагами.

Среда для обучения и публичной проверки задаётся следующими параметрами:

```python
import gymnasium as gym

env = gym.make(
    "LunarLander-v3",
    gravity=-10.5,
    enable_wind=True,
    wind_power=16.0,
    turbulence_power=1.6,
    max_episode_steps=600,
)
```

Цель — обучить агента, который получает наибольший средний доход за 100 эпизодов.

Можно использовать любые алгоритмы обучения с подкреплением.
